```python
# OTT Review Analysis Project
# Beginner-friendly Python project using the built-in csv module.

import csv

FILE_NAME = "ott_reviews_dataset.csv"


def load_reviews():
    """Read review records from the CSV file."""
    reviews = []

    try:
        with open(FILE_NAME, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            for row in reader:
                row["Rating"] = int(row["Rating"])
                reviews.append(row)

    except FileNotFoundError:
        print("Dataset file not found. Keep the CSV file in the same folder as this program.")

    return reviews


def show_summary(reviews):
    """Display basic information about the reviews."""
    if len(reviews) == 0:
        print("No review data available.")
        return

    total_rating = 0
    platform_count = {}
    genre_count = {}

    for review in reviews:
        total_rating += review["Rating"]

        platform = review["OTT_Platform"]
        genre = review["Genre"]

        platform_count[platform] = platform_count.get(platform, 0) + 1
        genre_count[genre] = genre_count.get(genre, 0) + 1

    average_rating = total_rating / len(reviews)

    print("\n========== OTT REVIEW ANALYSIS ==========")
    print("Total reviews:", len(reviews))
    print("Average rating:", round(average_rating, 2), "/ 5")

    print("\nReviews by OTT platform:")
    for platform, count in platform_count.items():
        print(platform, ":", count)

    print("\nReviews by genre:")
    for genre, count in genre_count.items():
        print(genre, ":", count)


def show_highest_rated(reviews):
    """Show the show with the highest individual review rating."""
    if not reviews:
        print("No review data available.")
        return

    highest = max(reviews, key=lambda item: item["Rating"])
    print("\nHighest-rated review:")
    print("Show:", highest["Show_Name"])
    print("Platform:", highest["OTT_Platform"])
    print("Rating:", highest["Rating"], "/ 5")
    print("Review:", highest["Review"])


def analyze_sentiment(reviews):
    """A very simple keyword-based sentiment check for beginners."""
    positive_words = ["amazing", "brilliant", "great", "good", "excellent",
                      "funny", "enjoyable", "beautiful", "entertaining", "interesting"]
    negative_words = ["boring", "slow", "weak", "disappointing", "confusing",
                      "stressful", "average", "difficult"]

    positive = 0
    negative = 0
    neutral = 0

    for review in reviews:
        words = review["Review"].lower()
        positive_found = any(word in words for word in positive_words)
        negative_found = any(word in words for word in negative_words)

        if positive_found and not negative_found:
            positive += 1
        elif negative_found and not positive_found:
            negative += 1
        else:
            neutral += 1

    print("\nSimple review sentiment:")
    print("Positive reviews:", positive)
    print("Negative reviews:", negative)
    print("Neutral / mixed reviews:", neutral)
    print("(This is a basic keyword method, not advanced AI sentiment analysis.)")


def search_by_platform(reviews):
    """Find reviews for a platform entered by the user."""
    platform_name = input("\nEnter OTT platform name: ").strip().lower()
    found = False

    for review in reviews:
        if platform_name in review["OTT_Platform"].lower():
            print("\nShow:", review["Show_Name"])
            print("Platform:", review["OTT_Platform"])
            print("Rating:", review["Rating"], "/ 5")
            print("Review:", review["Review"])
            found = True

    if not found:
        print("No reviews found for that platform.")


def main():
    reviews = load_reviews()

    if not reviews:
        return

    while True:
        print("\n----------- MAIN MENU -----------")
        print("1. Show overall analysis")
        print("2. Show highest-rated review")
        print("3. Analyze review sentiment")
        print("4. Search reviews by OTT platform")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()

        if choice == "1":
            show_summary(reviews)
        elif choice == "2":
            show_highest_rated(reviews)
        elif choice == "3":
            analyze_sentiment(reviews)
        elif choice == "4":
            search_by_platform(reviews)
        elif choice == "5":
            print("Thank you for using OTT Review Analysis!")
            break
        else:
            print("Invalid choice. Please enter a number from 1 to 5.")


if __name__ == "__main__":
    main()
```
